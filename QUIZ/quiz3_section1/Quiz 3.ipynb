(3) Recode a numeric column

<font size = "3">

- Import data from "races.csv" and assign it to a DataFrame
- Recode the "year" column into a new column "year_brackets"
with the following categories

$\qquad$ ``` ["1950-1965","1966-1980","1981-onwards"] ```

The easiest method is ``pd.cut()`` but you are not required to use this.

````python
# Your answer here

df3 = pd.read_csv("data_raw/races.csv")
bins = [1949, 1965, 1980, df3['year'].max()]
labels = ["1950-1965", "1966-1980", "1981-onwards"]

df3['year_brackets'] = pd.cut(df3['year'], bins=bins, labels=labels)

print(df3[['year', 'year_brackets']])
````

- To get full credit **ONLY** merge the "nationality" column from the secondary dataset (not all the columns)

````python
# Your answer here

df_sprint = pd.read_csv("data_raw/sprint_results.csv")
df_constructors = pd.read_csv("data_raw/constructors.csv")[['constructorId', 'nationality']]

df_merged = pd.merge(df_sprint, df_constructors, on='constructorId', how='left')

print(df_merged)
````
